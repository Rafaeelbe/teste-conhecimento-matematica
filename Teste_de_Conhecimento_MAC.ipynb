## Etapa 2 - Identificar as variáveis

### a) Qual é a variável independente?

A variável independente é o mês (`mes`), pois representa o tempo passando em meses normalmente.

### b) Qual é a variável dependente?

A variável dependente é o custo de transporte (`custo_transporte`), pois seu valor varia conforme o mês passa.

### Qual é a variável dependente?

- `mes`: representa o número do mês analisado.
- `custo_transporte`: representa o custo total de transporte da empresa naquele mês, em reais.

## Etapa 3 - Construir um gráfico

```python
plt.figure(figsize=(10, 5))

plt.scatter(df["mes"], df["custo_transporte"])

plt.title("Custos de Transporte por Mês")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.grid(True)

plt.show()
```

### Observação

O Custo de transporte cresce com o passar do tempo, mas não de maneira linear.

```python
a, b = np.polyfit(df["mes"], df["custo_transporte"], 1)

print("Coeficiente a:", a)
print("Coeficiente b:", b)
```

## Etapa 4 - Encontrar uma função

- a = 1791,61
- b = 42204,55
- f(x) = 1791,61x + 42204,55


### Cálculo da função

Foi utilizada uma regressão linear para encontrar a reta que melhor representa a tendência dos dados.

A função exigida foi uma função afim:

f(x) = ax + b

O coeficiente angular encontrado foi:

a ≈ 1791,61

O coeficiente linear encontrado foi:

b ≈ 42204,55

Assim:

f(x) = 1791,61x + 42204,55

## Etapa 5 - Interpretar a função

a) O que representa o coeficiente a no contexto do problema?

O coeficiente a é aproximadamente 1791,61. Ele representa a variação média estimada do custo de transporte a cada mês.

b) O que representa o coeficiente b?

O coeficiente b é aproximadamente 42204,55. Ele representa o valor que a reta estimaria para o custo quando x = 0. Como a contagem de meses começa a partir de 1, esse valor serve apenas de parâmetro para a reta.

c) O valor de a indica crescimento ou redução dos custos?

Como o valor de a é positivo, a função é crescente, ou seja, o valor do custo aumenta com o tempo.



## Etapa 6 - Representar a função graficamente

```python
df["previsao"] = a * df["mes"] + b
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    df["mes"],
    df["custo_transporte"],
    label="Dados reais"
)

plt.plot(
    df["mes"],
    df["previsao"],
    label="Função afim"
)

plt.title("Custos de Transporte e Reta de Tendência")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
meses_futuros = np.array([13, 14, 15])

previsoes = a * meses_futuros + b

for mes, valor in zip(meses_futuros, previsoes):
    print(f"Mês {mes}: R$ {valor:,.2f}")
```

## Etapa 7 - Fazer previsões

Utilizando a função:

f(x) = 1791,61x + 42204,55

foram realizadas as seguintes estimativas:

- Mês 13: f(13) = 1791,6(13) + 42204,55 ≈ 65.495,45
- Mês 14: f(14) = 1791,61(14) + 42204,55 ≈ 67.287,06
- Mês 15: f(15) = 1791,6(15) + 42204,55 ≈ 69.078,67

Os valores foram obtidos substituindo o número do mês na função encontrada.

## Etapa 8 - Análise dos resultados

### a) A função encontrada representa bem a tendência dos dados?

Relativamente sim, pois a função forma uma reta que passa bem próxima dos pontos dos dados reais, permitindo que se possa fazer previsões aproximadas a partir de uma média.

### b) Os custos estão aumentando ou diminuindo?

Os custos estão aumentando ao longo dos meses.

### c) Aproximadamente quanto o custo aumenta a cada mês?

De acordo com a função encontrada, a variação do custo (a) é de aproximadamente R$ 1.791,61 por mês.

## Etapa 9 - Conclusão

A análise dos dados dos 12 meses mostra que os custos de transporte apresentam uma tendência de crescimento ao longo do tempo. A regressão linear resultou na função aproximada f(x) = 1791,61x + 42204,55, indicando um aumento médio de aproximadamente R$ 1.791,61 por mês.

Com base nessa tendência, estima-se que os custos continuem aumentando nos próximos meses, chegando a aproximadamente RS 65.495,45 no mês 13, RS 67.287,06 no mês 14 e RS 69.078,67 no mês 15.
